## Checking the change in number of total loans and defaulted loans in each year

```python
total_loans = pd.pivot_table(loans, index = None, columns = 'year', values = 'loan_id' , aggfunc = 'count' )
defaulted_loans = pd.pivot_table(loans[loans['loan_status']=='Defaulted'], index = None, columns = 'year', values = 'loan_id', aggfunc = 'count')
total_loans.head()
defaulted_loans.head()
yoy_default = pd.concat([total_loans, defaulted_loans], axis = 0)
yoy_default.columns = ['2024','2025']
yoy_default.index = ['total_loans','default_loans']
yoy_default['perc_change'] = ((yoy_default['2025']/yoy_default['2024'] -1) * 100).round(3)
yoy_default
```

```python
# Checking the number of default loans under each risk band for both years alongwith their avg probability of default and avg number of previous late payments
collections[collections['collection_stage']=='Default'].groupby(['risk_band','year']).agg(
    {'collection_id':'count','probability_of_default':'mean','previous_late_payments':'mean'}).sort_values(
        by = ['year','risk_band'], key = lambda x: x.map({'Low':1,'Medium':2, 'High':3,'Very High': 4}))
```

### There seems some problem with risk band as low risk band has more number of defaults. Lets dig credit_risk table a little further

```python
credit_risk.groupby('risk_band').agg(
    avg_credit_score = ('credit_score','mean'),
    avg_risk_score = ('risk_score','mean'),
    avg_dti = ('debt_to_income','mean'),
    avg_lti = ('loan_to_income','mean'),
    avg_prob_of_default = ('probability_of_default','mean'),
    avg_late_payments = ('previous_late_payments','mean')
).sort_values(by = 'risk_band', key = lambda x : x.map({'Low':1,'Medium':2,'High': 3,'Very High': 4}))
```

### 1. Column 'risk_score' does not seem consistent with other columns

```python
#uploading the data to SQL server
import urllib.parse
from sqlalchemy import create_engine

# Database settings
host = r"localhost\SQLEXPRESS"
database = "finance_db"

# Build connection string for Windows Authentication
connection_string = (
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={host};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
)

params = urllib.parse.quote_plus(connection_string)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# ---------------------------------------------------------
# Upload your 3 DataFrames to SQL Server
# ---------------------------------------------------------

#  Upload Tables
customers.to_sql("customers", engine, if_exists="append", index=False)
credit_risk.to_sql("credit_risk", engine, if_exists="append", index=False)
loans.to_sql("loans", engine, if_exists="append", index=False)
collections.to_sql("collections", engine, if_exists="append", index=False)
repayments.to_sql("repayments", engine, if_exists="append", index=False)
acq.to_sql("acquisition", engine, if_exists="append", index=False)
app.to_sql("applications", engine, if_exists="append", index=False)
monthly_kpis.to_sql("monthly_kpis", engine, if_exists="append", index=False)


print("All tables uploaded successfully to 'finance_db'!")
```
